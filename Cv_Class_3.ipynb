**BGR to RGB**

```python
import cv2
import matplotlib.pyplot as plt

# Assuming 'img' holds the image data as indicated by global variables
img_bgr = img  # Assign the existing 'img' to 'img_bgr'

# Now 'img_bgr' should contain the image data
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb) # Display the converted image
plt.show()
```
